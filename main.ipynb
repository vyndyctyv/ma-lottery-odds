```python
import pandas as pd
import pull_ticket
import ticket_simulating

ticket_urls = pd.read_csv('ticket_urls.csv', header=0)

all_tickets = pull_ticket.pull_tickets(ticket_urls)
```

This program will calculate the odds of MA instant tickets (scratch cards). 

To see a table with the averages and simulated net value of 3 tickets, please scroll to the very bottom of this notebook.

When reading the dataframes of each ticket, please note that the "Prize" column contains a list of all the available prizes on a ticket with the cost of the ticket subtracted from the prize (hence the negative and zero value prizes).

Below each table, I have also included a sampling method to pull the average of 3 ticket pulls over 1000 instances. The logic is as follows:
* Your budget is 3x the value of the ticket. 
* As long as buying a ticket would not put you under 3x the value of the ticket and your net loss is equal to or under $0, you will continue buying tickets.
* If, at any point, your net gains exceed $0, you stop playing.

```python
index = 0 #Index for readability only
ticket_averages = []
for ticket in all_tickets:
    if ticket: #Check if ticket is valid. Functionality is not provided for tickets containing "winnings for life", as I do not believe that there is a universal value that I can put on that value.
        df = ticket[0]
        df_styled=df.style.set_caption(ticket[1])
        display(df_styled)
        weighted_average = sum(df['Prize'] * df['Odds']) / sum(df['Odds']) #Ticket Prize * Ticket Odds / sum of all odds
        print(f'Weighted average: ${round(weighted_average,2)}\n{ticket[2]}\nIndex: {index}\n')
        simulated_value = ticket_simulating.break_or_broke(ticket[0],10,1000) #3 tickets pulled using above logic in markdown; 1000 attempts
        ticket_averages.append([ticket[1],weighted_average,simulated_value])
        index+=1
ticket_df = pd.DataFrame(ticket_averages, columns=['Name', 'Average', 'Simulated Value'])
ticket_df.sort_values(by='Simulated Value',inplace=True)
styled_ticket_df = ticket_df.style.set_caption('All Ticket Averages')
display(styled_ticket_df)
```
